```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("points_all_parsed.csv", low_memory=False)

```

```python
unreturned_rate = df.groupby("Year")["is_unreturned"].mean()

plt.figure()
unreturned_rate.plot()
plt.ylabel("Unreturned Serve Rate")
plt.title("Unreturned Serves Over Time")
plt.show()
```
