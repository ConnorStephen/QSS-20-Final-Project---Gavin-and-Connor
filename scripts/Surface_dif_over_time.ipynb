```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("points_all_parsed.csv", low_memory=False)
```

```python
df["server_won"] = (
    ((df["Svr"] == 1) & (df["PtWinner"] == 1)) |
    ((df["Svr"] == 2) & (df["PtWinner"] == 2))
)

server_win_year = df.groupby("Year")["server_won"].mean()

plt.figure()
server_win_year.plot()
plt.ylabel("Server Win %")
plt.title("Server Dominance Over Time")
plt.show()
```

```python
surface_server = df.groupby(["Year", "Surface"])["server_won"].mean().unstack()
plt.figure()
surface_server.plot()
plt.ylabel("Server Win %")
plt.title("Server Win Rate by Surface Over Time")
plt.show()

```

```python
surface_diff = surface_server.max(axis=1) - surface_server.min(axis=1)

plt.figure()
surface_diff.plot()
plt.ylabel("Surface Difference in Server Win %")
plt.title("Different Surface Impact Over Time")
plt.show()
```
